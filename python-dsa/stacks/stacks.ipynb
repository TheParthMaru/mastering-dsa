# Stacks

- It is a linear data structure that follows the principle of Last In First Out (LIFO).

## Core stack operations

1. `push(x)`: Add `x` to the top.
2. `pop()`: Remove the top element.
3. `peek()` or `top()`: Looks at the top without remove it.
4. `is_empty()`: Check whether the stack is empty.
5. `size()`: Returns the number of elements.

## Stack vs a normal python list

- Stack restricts how you access the data.
- The whole point is that you are supposed to interact primarily with the top.
- A python list is not inherently a stack.
- For a list, you have random access and many operations.
- But you can use a list as a stack by restricting yourself to `append()` and `pop()`.



| Stack Operation | Python List       |
|-----------------|-------------------|
| `push(x)`       | `append(x)`       |
| `pop()`         | `pop()`           |
| `peek()`        | `stack[-1]`       |
| `is_empty()`    | `len(stack) == 0` |
| `size()`        | `len(stack)`      |

## Why append at the end?

- A good question is that why the `top` element is always the last or the most recent element?
- If we use a list to implement stack, we do we always append and not insert at 0th index?
- The reason is that technically it is possible but the operation is expensive and inefficient.
- If we insert at 0th index, list has to shift all the elements to the right to make space for the element to be inserted at 0th index.
- If we pop from 0th index, list has to shift all the elements to left to accomodate the empty space.
- The time to shift is `O(N)`.
- Whereas append to end and pop from end is `O(1)` or a better word is amortized complexity.

## A bit of knowledge on amortised complexity

- Amortized complexity is the average cost per operation over a long sequence of operations, not the cost of a single worst-case call.
- A few operations can be expensive, as long as they are rare enough that the average stays cheap.
- A Python list is a dynamic array.
- Most `appends` write into already-allocated space and are true `O(1)`. 
- When the array is full, Python allocates a bigger block (typically grows by a constant factor, often ~1.125x or similar) and copies every element. 
- That one append is `O(n)`.
- The `append` is cheap between growths.
- If you insert n elements:
    - Cheap appends: about `n` of them, each `O(1)`.
    - Copies: `1 + 2 + 4 + … + n ≈ 2n` element moves, which is `O(n)` total
- Total work for `n` appends: `O(n)`.
- Average per append: `O(n) / n = O(1)` amortized.
- `pop` is a true `O(1)`.

## Basic stack implementation

```python
stack = []

stack.append(5)
stack.append(10)
stack.append(15)

print(stack)
```

```python
# Top

print(stack[-1])
```

```python
# Pop

popped_element = stack.pop()

print(popped_element)
print(stack)
```

```python
# Size

print(len(stack))
```

```python
# Empty check (pythonic way)

if not stack:
    print(True)
else:
    print(False)
```

```python
# Empty check (generic way)

if len(stack) == 0:
    print(True)
else:
    print(False)
```
