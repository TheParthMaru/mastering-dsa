# Problems with Arrays

- Arrays take contiguous memory.
- If there is not enough contiguous memory, array creation becomes a problem.
- For arrays, contiguous memory is more important than the overall memory.
- Hence, at times, arrays can be memory in-efficient.
- To sort this problem, we have a Linked List which is a chain like data structure.

## Node

- It is the basic building block of a Linked List.
- Each node stores the data and the next.
- `data`: refers to the value (number, string).
- `next`: is a reference/pointer to the next node (`null` or `None` if it is the last node).
- When we chain multiple nodes together, we form a Linked List.
- The first node is called as a head and the last node is called as a tail.
- Linked list is a linear data structure.
- Linked list has no indexing.
- Individual elements can be accessed using traversing.

## Doubly Linked List

- Every node stores the address of the next node and the previous node.

## Types of Linked List

1. Singly linked list - Traverse only in a single direction.
2. Doubly linked list - Traverse both the direction.
3. Circular linked list - Extension of singly linked list where the tail node is connected to the head node instead of null. The list forms a cyclic formation.
4. Skip list - It has access to the next node but also additional pointers like next of next. Mostly an optimisation. Used for faster searching, jump nodes, etc.

## Coding a Singly Linked List

```python
# Creating the node class

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
# Creating the Linked List class

class LinkedList:

    # Creating an empty linked list
    def __init__(self):
        self.head = None
    
    def display(self):
        """
        1. Make a node that points to the head. This node represents the current node which will be used to traverse the linked list.
        2. Traverse until the current_node becomes None. This way we know we have reached the end.
        3. On every iteration, print the data which is stored in current_node.data.
        4. Make the current_node point to the next node.
        5. Print None in the end to show that end was reached.
        """

        current_node = self.head

        # Prefer truthiness rather than somevalue != None
        while current_node:
            print(current_node.data, "-> ", end="")
            current_node = current_node.next
        print(None)

    def prepend(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node

    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            return
        
        current_node = self.head
        while current_node.next is not None:
            current_node = current_node.next

        current_node.next = new_node


    def search(self, target):
        """
        1. Create a current node that points to the head.
        2. Loop until the current node becomes None.
        3. If the data in the current node matches the target value, return true.
        4. If no value was matched, return false after the loop ends.
        """

        current_node = self.head

        while current_node:
            if target == current_node.data:
                return True

            current_node = current_node.next
        
        return False

    def delete(self, target):
        # If the list is empty, just return
        if self.head is None:
            return False

        # If the head contains the target data
        if target == self.head.data:
            self.head = self.head.next
            return True

        current_node = self.head

        while current_node.next is not None:
            if current_node.next.data == target:
                current_node.next = current_node.next.next
                return True
            
            current_node = current_node.next

        return False

    
    def reverse(self):
        previous_node = None
        current_node = self.head

        while current_node:
            next_node = current_node.next
            current_node.next = previous_node
            previous_node = current_node
            current_node = next_node

        self.head = previous_node

    def find_middle(self):
        if not self.head:
            return None

        slow = self.head
        fast = self.head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        return slow.data
```

```python
ll = LinkedList()

ll.append(1)
ll.append(2)
ll.append(3)
ll.append(4)
# ll.append(5)

ll.display()
print(ll.find_middle())
```

## When to use a Linked List?

- Frequent insert/delete at the front (prepend) or delete head is O(1) whereas arrays needs shifting which takes O(N).
- When size grows unpredicatbly and you want to avoid resizing the array.
- When mutations at known pointer locations matter more than random access.

## When not to use a linked list?

- When indexing is required.
- When frequent scan and search is required.
- Use arrays or list unless you have a clear pointer-based mutation reasons.

## Doubly Linked List

- Each node has data, next, and prev.
- Can traverse both directions.
- Delete a given node in O(1) if you already have a reference to it (no need to find predecessor, prev is stored).
- Insert before/after a known node is also O(1).
- Cost: extra memory per node + more pointer updates (easier to bug).
- Use when: bidirectional navigation, LRU cache, editors/playlists where you move prev/next often, need O(1) remove by node reference.

Avoid when: you only ever walk forward and memory/simplicity matter, singly is enough.

## Circular Linked List

- Tail’s next points to head (not None). Forms a loop.
- Can be singly circular or doubly circular.
- No natural “end” — traversal must stop by counting nodes or returning to head.
- Use when: round-robin scheduling, multiplayer turns, playlist looping, Josephus-style cycling.
- Careful: easy infinite loops; display/search must detect “back to start.”
